```python
import numpy as np
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder

# 1. Expand data (Logistic Regression needs more than 2 samples to function)
x = ['input', 'input1', 'input', 'input1', 'input', 'input1']
y = ['output', 'output1', 'output', 'output1', 'output', 'output1']

# 2. Split data
X_train_raw, X_test_raw, Y_train_raw, Y_test_raw = train_test_split(x, y, train_size=0.8, random_state=25)

# 3. Encode Labels
le_x = LabelEncoder()
le_y = LabelEncoder()

# Fit on training, transform both
X_train = le_x.fit_transform(X_train_raw).reshape(-1, 1)
X_test = le_x.transform(X_test_raw).reshape(-1, 1)

Y_train = le_y.fit_transform(Y_train_raw)
Y_test = le_y.transform(Y_test_raw)

# 4. Train Model
model = LogisticRegression()
model.fit(X_train, Y_train)

# 5. Predict and Evaluate
y_pred = model.predict(X_test)
print(f"Accuracy: {accuracy_score(Y_test, y_pred)}")
print("Confusion Matrix:")
print(confusion_matrix(Y_test, y_pred))
```

```python
# encapsulation     

class Bank:
    def __init__(self,balance):
        self.__balance = balance
    def salary(self):
        return self._balance

account = Bank(1000)
print (account.salary())
```
