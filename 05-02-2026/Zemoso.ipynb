```python
from sklearn.feature_extraction.text import TfidfVectorizer 

para = ["The itsy bitsy spider crawled up the water spout. Down came the rain, and washed the spider out.Out came the sun, and dried up all the rain,and the itsy bitsy spider went up the spout again."]

vectorizer = TfidfVectorizer()
transform_vector = vectorizer.fit_transform(para)
print(vectorizer.get_feature_names_out())
print(transform_vector)
```

```python
from sklearn.feature_extraction.text import CountVectorizer 

para = ["The itsy bitsy spider crawled up the water spout. Down came the rain, and washed the spider out. Out came the sun, and dried up all the rain,and the itsy bitsy spider went up the spout again."]

vectorizer = CountVectorizer()
transform_vector = vectorizer.fit_transform(para)
print(vectorizer.get_feature_names_out())
print(transform_vector)
```

```python
mapping ={}

for i in " ".join(para).split():
    
    if i not in mapping:
        mapping[i] = 1
    else:
        mapping[i] +=1

print (mapping)
```
