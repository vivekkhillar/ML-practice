```python
from sklearn.feature_extraction.text import TfidfVectorizer

doc = ['My name is vivek',
' Vivek is a good boy']

vectorizer = TfidfVectorizer(stop_words='english')
x = vectorizer.fit_transform(doc)

print(vectorizer.get_feature_names_out())
print (x.toarray())
```

```python
from flask import Flask, jsonify, request
app = Flask(__name__)
@app.route('/',methods = ['GET'])
def home():
    return 'Hello vivek first API'

@app.route('/users',methods =['GET'])
def users():
    users_data = [{'id' : 1, 'name' : 'Vivek'},
             {'id': 2, 'name' :'Khillar'}]
    return Flask.jsonify(users_data)

@app.route('/users/<int:user_id>',methods =['GET'])
def user_id(user_id):
    users_data = [{'id' : 1, 'name' : 'Vivek'},
             {'id': 2, 'name' :'Khillar'}]
    for user in users_data:
        if user['id'] == user_id:
            return Flask.jsonify(user)
    return Flask.jsonify({'message':'Error found'})

@app.route('/users/returnbyid', methods=['POST']) 
def get_user_by_post():
    data = request.get_json()
    if not data:
        return jsonify({'message': 'No data provided'}), 400
    
    user_try_fetch = data.get("id")
    
    users_data = [{'id' : 1, 'name' : 'Vivek'},
             {'id': 2, 'name' :'Khillar'}]
    for user in users_data:
        if user['id'] == user_try_fetch:
            return jsonify(user)
    return jsonify({'message':'Error found'})

if __name__ == '__main__':
    # debug=True allows for auto-reloading when you change code
    app.run(port = 8080,debug=True, use_reloader=False)
```
