```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection   import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

```python
data = load_iris()
data
```

```python
x = data.data
y = data.target
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, random_state= 25, train_size= 0.7)
```

```python
pipeline = Pipeline([
    ('knn', KNeighborsClassifier(n_neighbors=5)),
])
```

```python
pipeline.fit(x_train,y_train)
```

```python
y_pred = pipeline.predict(x_test)
y_pred
```

```python
accuracy = accuracy_score(y_test,y_pred)
print(accuracy*100)
```

```python
# print which number not in list with maximum number
nums = [1,2,2,4]
mapping = {}
arr = []
for i in range (len(nums)):
    if nums[i] not in mapping:
        mapping[nums[i]] = 1
    else:
        mapping[nums[i]] +=1

arr.append(max(mapping,key = mapping.get))

for i in range(1,len(nums)+1):
    if i not in nums:
        arr.append(i)

arr
```

# Stack and POP

```python
target = [2,3,4]
lists = []
operation = []
n = 4


for i in range(1,n+1):
    print(i)
    if i in target:
        lists.append(i)
        operation.append("Push")
    elif target == lists:
        break
    elif i not in target:
        operation.append("Push")
        lists.append(i)
        operation.append("Pop")
        lists.remove(i)

print(lists)
operation


```
