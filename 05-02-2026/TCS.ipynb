## Reverse polish notation

```python
# Input: tokens = ["2","1","+","3","*"]
# Output: 9
# Explanation: ((2 + 1) * 3) = 9

stack =[]
tokens = ["4","13","5","/","+"]

for token in tokens:

    if token not in ('/*+-'):
        stack.append(int(token))

    else:
        b = stack.pop()
        a = stack.pop()

        if token  == '+':
            stack.append(a+b)
        elif token == '-':
            stack.append(a-b)
        elif token == '*':
            stack.append(a*b)
        elif token == '/':
            stack.append(a/b)

stack
```

```python
class Solution:
    def addStrings(self, num1: str, num2: str) -> str:
        i = len(num1) - 1
        j = len(num2) - 1
        carry = 0
        result = []

        while i >= 0 or j >= 0 or carry:
            digit1 = ord(num1[i]) - ord('0') if i >= 0 else 0
            digit2 = ord(num2[j]) - ord('0') if j >= 0 else 0

            total = digit1 + digit2 + carry
            carry = total // 10
            result.append(str(total % 10))

            i -= 1
            j -= 1

        return ''.join(reversed(result))
```

## Permutation of absolute difference between the strings

```python
class Solution:
    def findPermutationDifference(self, s: str, t: str) -> int:
        
        if len(s) != len(t):
            return 0
        sum = 0
        for ch in s:
            if ch in t:
                absolute_diff = abs(s.index(ch)-t.index(ch))
                sum += absolute_diff
        return sum
```

```python
Solution.findPermutationDifference("","abcde","edbac")
```

## Split a string in balanced string

```python
class Solution:
    def balancedStringSplit(self, s: str) -> int:
        
        balanced = 0
        count = 0
        current = ""
        res= []
        for i in s:
            current += i
            if i == "R":
                balanced+=1
            else:
                balanced-=1
            if balanced == 0:
                res.append(current)
                count+=1
        print (res,count)
```
