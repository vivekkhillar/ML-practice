# Convert the reviews to the NLP nltk which convert data to token and 

```python
import pandas as pd
import re
from nltk.tokenize import sent_tokenize,word_tokenize
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from warnings import filterwarnings
filterwarnings('ignore')
data = {'email_text': [
        "Get rich quick! Click here to win a million dollars.", # Spam
        "Meeting agenda for tomorrow. Please review the attached doc.", # Ham (Not Spam)
        "Congratulations! You are the lucky winner of a free iPhone.", # Spam
        "Hey, can we reschedule our lunch? I'm running late.", # Ham
        "URGENT: Your account verify password immediately." # Spam
    ],
    'label': [1, 0, 1, 0, 1] }# 1 = Spam, 0 = Not Spam}
df = pd.DataFrame(data)

def clean_text(data):
    text = re.sub(r'<.*?>','',data)
    text = re.sub(r'[^a-zA-Z\s]','',data)
    text = text.lower()
    return text

df['email_text'] = df['email_text'].apply(clean_text)

cv = CountVectorizer(max_features = 1000, stop_words= 'english', tokenizer = word_tokenize)
x= cv.fit_transform(df['email_text']).toarray()
y = df['label'].values

X_train,X_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 25)

model = GaussianNB()
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
print (accuracy_score(y_test,y_pred))


test_message = ['Meeting agenda for tomorrow']

text_message = [clean_text(test_message[0])]
transform = cv.transform(text_message).toarray()
prediction = model.predict(transform)
if prediction[0] == 1:
    print(f"Subject: '{text_message}' -> 🚨 SPAM")
else:
    print(f"Subject: '{text_message}' -> ✅ Safe")
```
